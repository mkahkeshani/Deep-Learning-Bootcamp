# 🎓 آزمایشگاه پرسپترون: پیش‌بینی قبولی دانشجو

## صورت مسئله

در این آزمایش می‌خواهیم یک پرسپترون ساده را بدون استفاده از هیچ کتابخانه‌ای و تنها با پایتون پیاده‌سازی کنیم تا پیش‌بینی کند که آیا یک دانشجو قبول می‌شود یا مردود.

برای هر دانشجو دو ویژگی در نظر می‌گیریم:
- **ساعت مطالعه:** تعداد ساعت مطالعه دانشجو
- **نمره آزمون قبلی:** نمره دانشجو از ۲۰

خروجی مدل به‌صورت دودویی تعریف می‌شود؛ مقدار `1` به معنی قبولی و مقدار `0` به معنی مردودی است.

در این آزمایش، مراحل محاسبه مجموع وزنی، اعمال تابع فعال‌ساز، محاسبه خطا و به‌روزرسانی وزن‌ها و بایاس را پیاده‌سازی خواهیم کرد.

</div>

<div dir="rtl" align="right">

## ۱. تعریف داده‌های آموزشی

در این مرحله، چند نمونه فرضی از اطلاعات دانشجویان را به‌عنوان داده‌های آموزشی تعریف می‌کنیم.

هر نمونه شامل ساعت مطالعه و نمره آزمون قبلی است. همچنین، برچسب واقعی هر دانشجو مشخص می‌کند که قبول شده است یا مردود.

این داده‌ها برای آموزش پرسپترون استفاده می‌شوند تا مدل بتواند رابطه‌ای میان ورودی‌ها و خروجی مورد انتظار یاد بگیرد.

</div>

```python
# Training data: [study hours, previous exam score]
X = [
    [1, 5],
    [2, 7],
    [3, 8],
    [4, 12],
    [5, 15],
    [6, 18],
    [1, 3],
    [2, 4]
]

# Labels: 1 = Pass, 0 = Fail
y = [0, 0, 0, 1, 1, 1, 0, 0]
```

<div dir="rtl" align="right">

## ۲. پیاده‌سازی و آموزش پرسپترون

در این مرحله، پرسپترون را از صفر پیاده‌سازی می‌کنیم.

ابتدا وزن‌ها، بایاس و نرخ یادگیری را تعیین می‌کنیم. سپس در هر مرحله، مجموع وزنی ورودی‌ها را محاسبه کرده و با استفاده از تابع پله‌ای، خروجی مدل را به دست می‌آوریم.

در ادامه، پیش‌بینی مدل را با برچسب واقعی مقایسه می‌کنیم. اگر مدل اشتباه کرده باشد، وزن‌ها و بایاس بر اساس قانون یادگیری پرسپترون اصلاح می‌شوند.

این فرآیند طی چند دوره آموزشی تکرار می‌شود.

</div>

```python
# Initial weights and bias
weights = [0.1, 0.1]
bias = -1

# Learning rate
learning_rate = 0.1

# Training process
for epoch in range(20):
    for i in range(len(X)):

        # Calculate weighted sum
        z = (
            X[i][0] * weights[0]
            + X[i][1] * weights[1]
            + bias
        )

        # Step activation function
        prediction = 1 if z >= 0 else 0

        # Calculate prediction error
        error = y[i] - prediction

        # Update weights and bias
        weights[0] += learning_rate * error * X[i][0]
        weights[1] += learning_rate * error * X[i][1]
        bias += learning_rate * error

print("Training completed!")
print("Weights:", weights)
print("Bias:", bias)
```

<div dir="rtl" align="right">

## ۳. پیش‌بینی برای یک دانشجوی جدید

پس از پایان آموزش، از پرسپترون برای پیش‌بینی وضعیت یک دانشجوی جدید استفاده می‌کنیم.

اطلاعات دانشجو به مدل داده می‌شود و پرسپترون با استفاده از وزن‌ها و بایاس یادگرفته‌شده، مجموع وزنی را محاسبه می‌کند.

در نهایت، تابع فعال‌ساز خروجی مدل را مشخص می‌کند:

- خروجی `1`: پیش‌بینی قبولی
- خروجی `0`: پیش‌بینی مردودی

در این مرحله، مدل فقط پیش‌بینی می‌کند و از پاسخ واقعی دانشجوی جدید برای به‌روزرسانی وزن‌ها استفاده نمی‌کنیم.

</div>

```python
# New student: 4 study hours and previous score of 14
student = [4, 14]

# Calculate weighted sum
z = (
    student[0] * weights[0]
    + student[1] * weights[1]
    + bias
)

# Apply activation function
prediction = 1 if z >= 0 else 0

if prediction == 1:
    print("Prediction: Pass")
else:
    print("Prediction: Fail")
```
